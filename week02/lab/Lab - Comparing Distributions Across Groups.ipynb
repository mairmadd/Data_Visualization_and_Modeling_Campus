### Step 3: Evidence

Create a **histogram comparison and a box-plot comparison** for this quantitative feature using the same observations and group definitions. A violin plot is optional. Use descriptive titles, units, readable labels, a consistent group order, and named colors. Keep group colors consistent across plots. Use shared histogram bins; use whole-number bins when appropriate for discrete counts. Label whether the vertical axis shows counts or another documented scale.

Start with the histogram and estimate the distributions visually. Then calculate group summaries and use the box plot to compare them. Retain valid extreme observations; do not remove them just to make figures look cleaner.


```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

colors = {4: "blue", 6: "orange", 8: "green"}

plt.figure(figsize=(10, 6))

for cylinders in [4, 6, 8]: 
    data = investigation1_df[
        investigation1_df["cylinders"] == cylinders 
        ]["mpg"] 
    plt.hist(
            data,
            bins=np.arange(5, 51, 2),
            alpha=0.4, 
            label=f"{cylinders} cylinders", 
            color=colors[cylinders]
     )

plt.title("Fuel Efficiency Distribution by Cylinder Count")
plt.xlabel("Fuel Efficiency (mpg)")
plt.ylabel("Number of Vehicles")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

**From the histogram, estimate each group's typical values and spread. Describe shape, overlap, and unusual observations. How do group sizes affect the comparison?**

Your answer:

The histogram shows that 4-cylinder vehicles generally have the highest fuel efficiency, with most values between approximately 23 and 35 mpg. Six-cylinder vehicles are mostly between 15 and 22 mpg, while 8-cylinder vehicles are concentrated around 12 to 18 mpg. The 4-cylinder group has the widest spread and a longer right tail, reaching approximately 47 mpg. The groups overlap, especially between 6- and 8-cylinder vehicles. The 4-cylinder group contains 204 vehicles, compared with 84 six-cylinder and 103 eight-cylinder vehicles. Because the group sizes differ, the histogram bar heights should be compared carefully. Overall, fuel efficiency tends to decrease as cylinder count increases.

```python
# Display grouped .describe() summaries and calculate or display the IQR for each group.
summary = investigation1_df.groupby('cylinders')['mpg'].describe()

summary['IQR'] = summary['75%'] - summary['25%']

display(summary.round(2))
```

```python
# Create and label the box-plot comparison.
plt.figure(figsize=(9, 6))

sns.boxplot(
data=investigation1_df,
x="cylinders",
y="mpg",
order=[4, 6, 8],
hue="cylinders",
palette={4: "blue", 6: "orange", 8: "green"},
legend=False
)

plt.title("Fuel Efficiency by Cylinder Count")
plt.xlabel("Number of Cylinders")
plt.ylabel("Fuel Efficiency (mpg)")
plt.grid(axis="y", alpha=0.2)
plt.show()
```

**How do the calculated summaries compare with your visual estimates? Which groups look similar or different? Cite centers and spreads with units, and explain what each plot reveals or hides.**

The calculated statistics support my visual estimates. Four-cylinder vehicles have the highest median fuel efficiency at 28.25 mpg, followed by six-cylinder vehicles at 19 mpg and eight-cylinder vehicles at 14 mpg. The 4-cylinder group also has the largest interquartile range at 8 mpg, compared with 3 mpg for both the 6- and 8-cylinder groups. The 6- and 8-cylinder groups have more similar spreads, although their medians differ. The histogram shows the distribution shapes, frequencies, and overlap, while the box plot makes it easier to compare medians, quartiles, variability, and potential outliers. Both plots show that fuel efficiency generally decreases as cylinder count increases.
